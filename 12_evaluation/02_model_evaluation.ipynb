## 🔬 Model Comparison

```python
class ModelEvaluator:
    def __init__(self, eval_dataset):
        self.eval_dataset = eval_dataset
    
    def evaluate_models(self, models: list):
        """Compare multiple models."""
        
        results = {}
        
        for model in models:
            scores = []
            costs = []
            latencies = []
            
            for item in self.eval_dataset:
                start = time.time()
                
                response = client.chat.completions.create(
                    model=model,
                    messages=[{"role": "user", "content": item["input"]}]
                )
                
                latency = time.time() - start
                
                # Calculate score
                score = f1_score(
                    response.choices[0].message.content,
                    item["expected"]
                )
                
                # Calculate cost
                cost = calculate_cost(model, response.usage)
                
                scores.append(score)
                costs.append(cost)
                latencies.append(latency)
            
            results[model] = {
                'avg_score': np.mean(scores),
                'avg_cost': np.mean(costs),
                'avg_latency': np.mean(latencies),
                'total_cost': np.sum(costs)
            }
        
        return results
```

## 💰 Cost vs Quality Tradeoff

```python
# Decision matrix
def choose_model(task_complexity: str, budget: str):
    """Choose best model for use case."""
    
    if budget == "low":
        if task_complexity == "simple":
            return "gpt-3.5-turbo"  # Cheap, good enough
        else:
            return "gpt-3.5-turbo"  # Still cheap
    
    elif budget == "medium":
        if task_complexity == "simple":
            return "gpt-3.5-turbo"  # Overkill to use GPT-4
        else:
            return "gpt-4"  # Worth it for complex tasks
    
    else:  # high budget
        return "gpt-4"  # Best quality
```

## ✅ Summary

**Model selection criteria:**
1. **Quality**: How accurate?
2. **Cost**: Can you afford it?
3. **Latency**: Fast enough?
4. **Reliability**: Uptime, rate limits

**Best practices:**
- Evaluate on YOUR data
- Consider total cost (not just per-token)
- Test at scale
- Monitor in production
