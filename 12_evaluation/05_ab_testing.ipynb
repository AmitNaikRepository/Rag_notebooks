## 🧪 A/B Test Framework

```python
import random
from typing import Dict

class ABTest:
    def __init__(self, name: str, variants: dict):
        """
        Args:
            name: Test name
            variants: {"control": config, "treatment": config}
        """
        self.name = name
        self.variants = variants
        self.results = {v: [] for v in variants}
    
    def assign_variant(self, user_id: str) -> str:
        """Consistently assign user to variant."""
        # Use hash for consistent assignment
        hash_val = int(hashlib.md5(
            f"{self.name}:{user_id}".encode()
        ).hexdigest(), 16)
        
        return "control" if hash_val % 2 == 0 else "treatment"
    
    def record_result(self, variant: str, metric: float):
        """Record metric for variant."""
        self.results[variant].append(metric)
    
    def get_results(self) -> Dict:
        """Calculate results."""
        return {
            variant: {
                'mean': np.mean(values),
                'std': np.std(values),
                'count': len(values)
            }
            for variant, values in self.results.items()
        }
```

## 📊 Statistical Significance

```python
from scipy import stats

def is_significant(control: list, treatment: list, alpha=0.05):
    """Check if difference is statistically significant."""
    
    # T-test
    t_stat, p_value = stats.ttest_ind(control, treatment)
    
    is_sig = p_value < alpha
    
    return {
        'p_value': p_value,
        'is_significant': is_sig,
        'control_mean': np.mean(control),
        'treatment_mean': np.mean(treatment),
        'difference': np.mean(treatment) - np.mean(control),
        'relative_change': (
            (np.mean(treatment) - np.mean(control)) / 
            np.mean(control) * 100
        )
    }

# Example
control_scores = [4.1, 4.3, 4.0, 4.2, 4.1]
treatment_scores = [4.5, 4.6, 4.4, 4.7, 4.5]

result = is_significant(control_scores, treatment_scores)
print(f"Significant: {result['is_significant']}")
print(f"Improvement: {result['relative_change']:.1f}%")
```

### Sample Size Calculator:
```python
def required_sample_size(
    baseline: float,
    min_detectable_effect: float,
    power: float = 0.8,
    alpha: float = 0.05
):
    """Calculate required sample size."""
    # Simplified formula
    effect_size = min_detectable_effect / baseline
    n = (16 * (1 - power)) / (effect_size ** 2)
    return int(n)

# Example: Detect 5% improvement
n = required_sample_size(
    baseline=4.0,
    min_detectable_effect=0.2  # 5% of 4.0
)
print(f"Need {n} samples per variant")
```

## ✅ Summary

**A/B testing best practices:**

1. **Define hypothesis** - What are you testing?
2. **Choose metric** - How to measure success?
3. **Calculate sample size** - How many users?
4. **Run test** - Consistent assignment
5. **Analyze results** - Statistical significance
6. **Make decision** - Ship or rollback

**Common tests:**
- Model comparison (GPT-3.5 vs GPT-4)
- Prompt variations
- Temperature settings
- System prompts
- Feature additions

**Gotchas:**
- Don't stop test early
- Account for novelty effect
- Watch for interaction effects
- Monitor multiple metrics
