````python
from langchain.text_splitter import (
    RecursiveCharacterTextSplitter,
    CharacterTextSplitter
)
from typing import List

print("✅ Setup complete")
````

## 📖 Why Chunk?

**Problem:**
- LLMs have token limits (4K-32K)
- Embeddings work best on focused text
- Need to retrieve only relevant parts

**Solution:**
- Split documents into smaller pieces
- Each chunk is searchable
- Return only relevant chunks

**Chunk Size Guidelines:**
```
100-200 chars:  Very specific search
300-500 chars:  Balanced (recommended)
1000+ chars:    More context, less precise
```

## 🔪 Strategy 1: Fixed-Size Chunking

````python
text = """Machine learning is amazing. It can do many things. 
Neural networks are powerful. They learn from data. 
Deep learning is a subset. It uses many layers."""

splitter = CharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=10,
    separator=" "
)

chunks = splitter.split_text(text)

print(f"Created {len(chunks)} chunks:\n")
for i, chunk in enumerate(chunks, 1):
    print(f"Chunk {i} ({len(chunk)} chars): {chunk}")
    print()
````

## 📝 Strategy 2: Recursive Chunking (Best!)

````python
# Tries to split on: paragraphs → sentences → words → characters
splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    separators=["\n\n", "\n", ". ", " ", ""]
)

long_text = """# Introduction

Machine learning is a fascinating field. It has many applications.

## Deep Learning

Deep learning uses neural networks. These networks have multiple layers. 
They can learn complex patterns from data."""

chunks = splitter.split_text(long_text)

print(f"Created {len(chunks)} chunks:\n")
for i, chunk in enumerate(chunks, 1):
    print(f"Chunk {i}:")
    print(chunk)
    print("---")
````

## 🔄 Chunk Overlap Explained

**Without overlap:**
```
Chunk 1: "The cat sat on..."
Chunk 2: "...the mat happily"
Problem: Context lost between chunks!
```

**With overlap:**
```
Chunk 1: "The cat sat on the mat"
Chunk 2: "sat on the mat happily"  (overlap!)
Benefit: Context preserved!
```

**Recommendation:** 10-20% overlap

## ✅ Summary

### Chunking Strategies:
1. **Fixed-size**: Simple, predictable
2. **Recursive**: Smart, context-aware (BEST!)
3. **Semantic**: Group by meaning

### Best Practices:
- 📏 **Size**: 300-500 chars typical
- 🔄 **Overlap**: 10-20% of chunk size
- 📝 **Separators**: Paragraph → Sentence → Word
- 🎯 **Test**: Try different sizes, measure quality
