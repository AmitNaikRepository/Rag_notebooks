## 📑 Loading Plain Text

```python
def load_text_file(file_path: str) -> str:
    """Load plain text file."""
    with open(file_path, 'r', encoding='utf-8') as f:
        return f.read()

# Create sample file
sample_text = """This is a sample document.
It contains multiple paragraphs.

This is useful for RAG systems."""

with open('sample.txt', 'w') as f:
    f.write(sample_text)

# Load it
text = load_text_file('sample.txt')
print(f"Loaded {len(text)} characters")
print(f"\nContent:\n{text}")
```

## 📄 Loading PDFs

```python
from pypdf import PdfReader

def load_pdf(file_path: str) -> List[Dict[str, any]]:
    """Load PDF and return pages with metadata."""
    reader = PdfReader(file_path)
    
    pages = []
    for i, page in enumerate(reader.pages):
        pages.append({
            'page_number': i + 1,
            'text': page.extract_text(),
            'metadata': {
                'source': file_path,
                'page': i + 1,
                'total_pages': len(reader.pages)
            }
        })
    
    return pages

# Example usage (if PDF exists)
# pages = load_pdf('document.pdf')
# print(f"Loaded {len(pages)} pages")
print("PDF loader ready ✅")
```

## 📝 Loading DOCX

```python
from docx import Document

def load_docx(file_path: str) -> str:
    """Load DOCX file."""
    doc = Document(file_path)
    paragraphs = [p.text for p in doc.paragraphs]
    return '\n'.join(paragraphs)

# Example
# text = load_docx('document.docx')
print("DOCX loader ready ✅")
```

## 🌐 Loading HTML/Markdown

```python
from bs4 import BeautifulSoup
import markdown

def load_html(html_string: str) -> str:
    """Extract text from HTML."""
    soup = BeautifulSoup(html_string, 'html.parser')
    return soup.get_text()

def load_markdown(md_string: str) -> str:
    """Convert markdown to plain text."""
    html = markdown.markdown(md_string)
    return load_html(html)

# Test
html = "<h1>Title</h1><p>Content here</p>"
text = load_html(html)
print(f"HTML → Text: {text}")

md = "# Title\n\nContent **here**"
text = load_markdown(md)
print(f"Markdown → Text: {text}")
```

## 🔧 Universal Document Loader

```python
class DocumentLoader:
    """Universal document loader."""
    
    def load(self, file_path: str) -> Dict[str, any]:
        """Load any supported document type."""
        ext = os.path.splitext(file_path)[1].lower()
        
        loaders = {
            '.txt': self._load_text,
            '.pdf': self._load_pdf,
            '.docx': self._load_docx,
            '.md': self._load_markdown,
        }
        
        if ext not in loaders:
            raise ValueError(f"Unsupported file type: {ext}")
        
        return loaders[ext](file_path)
    
    def _load_text(self, path):
        with open(path, 'r', encoding='utf-8') as f:
            return {'text': f.read(), 'source': path}
    
    def _load_pdf(self, path):
        pages = load_pdf(path)
        text = '\n'.join(p['text'] for p in pages)
        return {'text': text, 'source': path, 'pages': len(pages)}
    
    def _load_docx(self, path):
        return {'text': load_docx(path), 'source': path}
    
    def _load_markdown(self, path):
        with open(path, 'r', encoding='utf-8') as f:
            return {'text': load_markdown(f.read()), 'source': path}

# Test
loader = DocumentLoader()
doc = loader.load('sample.txt')
print(f"Loaded: {doc['source']}")
print(f"Length: {len(doc['text'])} chars")
```
