````python
import os
from dotenv import load_dotenv
from groq import Groq

load_dotenv()
client = Groq(api_key=os.getenv('GROQ_API_KEY'))

def call_llm(prompt, **params):
    response = client.chat.completions.create(
        model="mixtral-8x7b-32768",
        messages=[{"role": "user", "content": prompt}],
        **params
    )
    return response.choices[0].message.content

print("✅ Setup complete")
````

## 🌡️ Temperature: Control Randomness

**Range:** 0.0 to 2.0

**What it does:**
- Low (0.0-0.3): Deterministic, focused
- Medium (0.5-0.8): Balanced
- High (1.0-2.0): Creative, varied

**Use cases:**
```
0.0 - Code generation, classification, factual Q&A
0.3 - Customer support, documentation
0.7 - General chatbot, content creation
1.0+ - Creative writing, brainstorming
```

````python
prompt = "Complete: The future of AI is"

temps = [0.0, 0.5, 1.0, 1.5]

print("Temperature Comparison:\n" + "="*60)
for temp in temps:
    response = call_llm(prompt, temperature=temp, max_tokens=30)
    print(f"\nTemp={temp}: {response}")
````

## 🎯 top_p (Nucleus Sampling)

**Range:** 0.0 to 1.0

**What it does:**
- Considers only tokens that sum to top_p probability
- Alternative to temperature
- More consistent output variance

**Recommendation:**
- Use EITHER temperature OR top_p (not both)
- top_p=0.9 is good default
- top_p=0.1 for very focused output

````python
print("top_p Comparison:\n" + "="*60)
for p in [0.1, 0.5, 0.9, 1.0]:
    response = call_llm(
        "List 3 colors:",
        temperature=1.0,  # High temp to see effect
        top_p=p,
        max_tokens=20
    )
    print(f"\ntop_p={p}: {response}")
````

## 🚫 Penalties: Reduce Repetition

### frequency_penalty
- Penalizes tokens based on how often they appear
- Range: -2.0 to 2.0
- Positive = less repetition

### presence_penalty
- Penalizes tokens that already appeared (any amount)
- Range: -2.0 to 2.0
- Positive = encourages new topics

````python
prompt = "Write about dogs. " * 3  # Encourage repetition

print("No Penalty:")
print(call_llm(prompt, max_tokens=50))

print("\nWith frequency_penalty=1.0:")
print(call_llm(prompt, frequency_penalty=1.0, max_tokens=50))
````

## ✋ Stop Sequences

**What:** String(s) that stop generation
**Use:** Control output format

````python
# Stop at newline (single-line response)
response = call_llm(
    "Write a haiku about code",
    stop=["\n\n"],  # Stop at double newline
    max_tokens=100
)
print("Response (stops at double newline):")
print(response)
````

## 📊 Parameter Combinations Cheat Sheet

````python
PARAMETER_PRESETS = {
    'code_generation': {
        'temperature': 0.0,
        'max_tokens': 500,
        'top_p': 1.0,
    },
    'classification': {
        'temperature': 0.0,
        'max_tokens': 10,
        'top_p': 1.0,
    },
    'creative_writing': {
        'temperature': 0.9,
        'max_tokens': 1000,
        'presence_penalty': 0.6,
    },
    'chatbot': {
        'temperature': 0.7,
        'max_tokens': 150,
        'frequency_penalty': 0.3,
    },
    'summarization': {
        'temperature': 0.3,
        'max_tokens': 200,
    },
}

# Example: Use preset
response = call_llm(
    "Classify: 'I love this!'",
    **PARAMETER_PRESETS['classification']
)
print(f"Classification result: {response}")
````

## ✅ Summary

### Key Parameters:
- **temperature**: Creativity (0=focused, 2=random)
- **max_tokens**: Output length limit
- **top_p**: Alternative to temperature
- **frequency_penalty**: Reduce repetition
- **presence_penalty**: Encourage new topics
- **stop**: Control output boundaries

### Quick Rules:
1. Start with defaults
2. Adjust one parameter at a time
3. Test on multiple examples
4. Use presets for common tasks
