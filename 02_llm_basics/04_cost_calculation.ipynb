````python
from typing import Dict, List
from datetime import datetime, timedelta
import json

# Latest pricing (check provider sites for updates)
PRICING = {
    'gpt-4': {'input': 0.03, 'output': 0.06},
    'gpt-3.5-turbo': {'input': 0.0015, 'output': 0.002},
    'claude-3-opus': {'input': 0.015, 'output': 0.075},
    'claude-3-sonnet': {'input': 0.003, 'output': 0.015},
    'groq-mixtral': {'input': 0.0, 'output': 0.0},  # Free tier
}

print("✅ Pricing data loaded")
````

## 💵 Cost Calculator

````python
class CostCalculator:
    """Calculate and track LLM API costs."""
    
    def __init__(self, pricing: Dict = PRICING):
        self.pricing = pricing
        self.usage_log = []
    
    def calculate_cost(self, model: str, input_tokens: int, output_tokens: int) -> float:
        """Calculate cost for a single API call."""
        if model not in self.pricing:
            raise ValueError(f"Unknown model: {model}")
        
        prices = self.pricing[model]
        cost = (input_tokens / 1000 * prices['input'] + 
                output_tokens / 1000 * prices['output'])
        return cost
    
    def log_usage(self, model: str, input_tokens: int, output_tokens: int, 
                  metadata: Dict = None):
        """Log an API call."""
        cost = self.calculate_cost(model, input_tokens, output_tokens)
        
        entry = {
            'timestamp': datetime.now().isoformat(),
            'model': model,
            'input_tokens': input_tokens,
            'output_tokens': output_tokens,
            'cost': cost,
            'metadata': metadata or {}
        }
        self.usage_log.append(entry)
        return cost
    
    def get_stats(self) -> Dict:
        """Get usage statistics."""
        if not self.usage_log:
            return {'total_cost': 0, 'total_calls': 0}
        
        total_cost = sum(entry['cost'] for entry in self.usage_log)
        total_tokens = sum(
            entry['input_tokens'] + entry['output_tokens']
            for entry in self.usage_log
        )
        
        return {
            'total_cost': total_cost,
            'total_calls': len(self.usage_log),
            'total_tokens': total_tokens,
            'avg_cost_per_call': total_cost / len(self.usage_log),
        }

# Example usage
calc = CostCalculator()

# Log some API calls
calc.log_usage('gpt-3.5-turbo', 100, 50)
calc.log_usage('gpt-4', 200, 150)
calc.log_usage('groq-mixtral', 500, 300)

# Get statistics
stats = calc.get_stats()
print("Usage Statistics:")
print(f"  Total cost: ${stats['total_cost']:.4f}")
print(f"  Total calls: {stats['total_calls']}")
print(f"  Avg cost/call: ${stats['avg_cost_per_call']:.4f}")
````

## 📊 Budget Estimator

````python
def estimate_monthly_cost(
    calls_per_day: int,
    avg_input_tokens: int,
    avg_output_tokens: int,
    model: str
) -> Dict:
    """Estimate monthly costs."""
    calc = CostCalculator()
    cost_per_call = calc.calculate_cost(model, avg_input_tokens, avg_output_tokens)
    
    daily_cost = cost_per_call * calls_per_day
    monthly_cost = daily_cost * 30
    yearly_cost = monthly_cost * 12
    
    return {
        'cost_per_call': cost_per_call,
        'daily_cost': daily_cost,
        'monthly_cost': monthly_cost,
        'yearly_cost': yearly_cost
    }

# Example: Chatbot scenario
estimate = estimate_monthly_cost(
    calls_per_day=1000,
    avg_input_tokens=100,
    avg_output_tokens=200,
    model='gpt-3.5-turbo'
)

print("Chatbot Cost Estimate (GPT-3.5):")
print(f"  Per call: ${estimate['cost_per_call']:.6f}")
print(f"  Daily: ${estimate['daily_cost']:.2f}")
print(f"  Monthly: ${estimate['monthly_cost']:.2f}")
print(f"  Yearly: ${estimate['yearly_cost']:.2f}")
````
