## 🔒 Security Best Practices

### ❌ NEVER Do This:
```python
api_key = "sk-1234567890"  # Hardcoded!
```

### ✅ Always Do This:
```python
api_key = os.getenv("OPENAI_API_KEY")  # From .env
```

````python
class APIKeyManager:
    """Secure API key management."""
    
    def __init__(self, env_file: str = ".env"):
        load_dotenv(env_file)
        self.keys = self._load_keys()
    
    def _load_keys(self) -> Dict[str, str]:
        """Load all API keys."""
        return {
            'openai': os.getenv('OPENAI_API_KEY'),
            'anthropic': os.getenv('ANTHROPIC_API_KEY'),
            'groq': os.getenv('GROQ_API_KEY'),
        }
    
    def validate(self) -> Dict[str, bool]:
        """Check which keys are configured."""
        return {
            provider: (key is not None and not key.startswith('your-'))
            for provider, key in self.keys.items()
        }
    
    def get_key(self, provider: str) -> Optional[str]:
        """Safely get API key."""
        key = self.keys.get(provider)
        if not key or key.startswith('your-'):
            raise ValueError(f"API key for {provider} not configured")
        return key

# Test
manager = APIKeyManager()
status = manager.validate()

print("API Key Status:")
for provider, configured in status.items():
    print(f"  {provider}: {'✅ Configured' if configured else '❌ Missing'}")
````

## 💰 Cost Monitoring

Set up budget alerts to avoid surprises!

````python
class CostTracker:
    """Track API usage costs."""
    
    def __init__(self, budget_limit: float = 100.0):
        self.budget_limit = budget_limit
        self.total_cost = 0.0
        self.usage_log = []
    
    def add_cost(self, provider: str, cost: float, tokens: int):
        """Record API call cost."""
        self.total_cost += cost
        self.usage_log.append({
            'timestamp': datetime.now().isoformat(),
            'provider': provider,
            'cost': cost,
            'tokens': tokens
        })
        
        # Check budget
        if self.total_cost > self.budget_limit:
            print(f"⚠️ WARNING: Budget exceeded! ${self.total_cost:.2f} > ${self.budget_limit:.2f}")
    
    def get_summary(self) -> Dict:
        """Get cost summary."""
        return {
            'total_cost': self.total_cost,
            'budget_remaining': self.budget_limit - self.total_cost,
            'total_calls': len(self.usage_log)
        }

# Example
tracker = CostTracker(budget_limit=50.0)
tracker.add_cost('openai', 0.05, 1000)
tracker.add_cost('anthropic', 0.02, 500)

summary = tracker.get_summary()
print(f"\nCost Summary:")
print(f"  Total: ${summary['total_cost']:.2f}")
print(f"  Remaining: ${summary['budget_remaining']:.2f}")
print(f"  Calls: {summary['total_calls']}")
````
