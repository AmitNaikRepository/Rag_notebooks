## 📖 Why Environment Setup Matters

Many AI engineering beginners jump straight into code without proper environment setup. This leads to:
- **Dependency conflicts** - Different projects requiring different versions
- **Security risks** - API keys hardcoded in notebooks
- **Reproducibility issues** - "Works on my machine" syndrome
- **Cost overruns** - No monitoring or limits set up

### The Solution
Professional AI engineers use:
- **Virtual environments** - Isolated Python environments per project
- **Environment variables** - Secure credential management
- **Requirements files** - Reproducible dependency management
- **Version control** - Git with proper .gitignore

### Real-World Context
In production environments, you'll often work with:
- Multiple projects with different dependencies
- Team members who need to reproduce your environment
- CI/CD pipelines that need consistent setups
- Security audits that check for exposed credentials

**Getting this right from day one saves countless hours of debugging later!**

### 💡 Understanding Virtual Environments

**What is a virtual environment?**
Think of it as a separate "bubble" for each Python project. Each bubble has its own:
- Python interpreter
- Installed packages
- Package versions

**Why use them?**
```
Project A needs: pandas 1.5.0, numpy 1.24.0
Project B needs: pandas 2.0.0, numpy 1.26.0

Without venv: 💥 Conflict!
With venv: ✅ Both work independently
```

**How to create one:**
```bash
# Create virtual environment
python -m venv venv

# Activate it
# Linux/Mac:
source venv/bin/activate

# Windows:
venv\Scripts\activate
```

**Pro Tip:** If you ran `setup.sh`, this is already done for you!

## 📦 Step 2: Verify Package Installation

Let's check that all required packages are installed correctly.

````python
# List of core packages we need
REQUIRED_PACKAGES = [
    ('openai', 'OpenAI API client'),
    ('anthropic', 'Anthropic Claude API client'),
    ('groq', 'Groq API client'),
    ('langchain', 'LLM application framework'),
    ('chromadb', 'Vector database'),
    ('pandas', 'Data manipulation'),
    ('numpy', 'Numerical computing'),
    ('fastapi', 'API framework'),
    ('jupyter', 'Jupyter notebooks'),
    ('python-dotenv', 'Environment variables'),
]

print("📦 Checking Required Packages")
print("=" * 70)

missing_packages = []

for package_name, description in REQUIRED_PACKAGES:
    try:
        # Try to import the package
        package = __import__(package_name.replace('-', '_'))
        
        # Get version if available
        version = getattr(package, '__version__', 'unknown')
        
        print(f"✅ {package_name:<20} {version:<15} - {description}")
    except ImportError:
        print(f"❌ {package_name:<20} {'NOT INSTALLED':<15} - {description}")
        missing_packages.append(package_name)

print("=" * 70)

if missing_packages:
    print(f"\n⚠️  Missing packages: {', '.join(missing_packages)}")
    print("\nTo install missing packages, run:")
    print(f"pip install {' '.join(missing_packages)}")
else:
    print("\n🎉 All required packages are installed!")
````

### 📊 Visualize Package Dependencies

Let's see how many packages we actually have installed:

````python
import subprocess
import json

# Get list of installed packages
result = subprocess.run(
    ['pip', 'list', '--format=json'],
    capture_output=True,
    text=True
)

packages = json.loads(result.stdout)
total_packages = len(packages)

print(f"📊 Environment Statistics")
print("=" * 50)
print(f"Total installed packages: {total_packages}")
print(f"Core AI/LLM packages: {len(REQUIRED_PACKAGES)}")
print(f"Supporting packages: {total_packages - len(REQUIRED_PACKAGES)}")
print("=" * 50)

# Show a few interesting packages
print("\n🔍 Some installed packages:")
for pkg in packages[:10]:
    print(f"  • {pkg['name']:<25} {pkg['version']}")
print(f"  ... and {total_packages - 10} more")
````

## 🔐 Step 3: Configure API Keys Securely

### ⚠️ CRITICAL: Never Hardcode API Keys!

**WRONG (Security Risk):**
```python
# ❌ DON'T DO THIS!
api_key = "sk-1234567890abcdef"  # This can be committed to Git!
```

**RIGHT (Secure):**
```python
# ✅ DO THIS!
import os
api_key = os.getenv("OPENAI_API_KEY")  # Loaded from .env file
```

### How to Get API Keys

1. **OpenAI** (GPT-4, GPT-3.5)
   - Go to: https://platform.openai.com/api-keys
   - Click "Create new secret key"
   - Cost: ~$0.03 per 1K tokens (GPT-4)

2. **Anthropic** (Claude)
   - Go to: https://console.anthropic.com/
   - Generate API key
   - Cost: ~$0.003 per 1K tokens (Claude 3.5 Sonnet)

3. **Groq** (Fast inference)
   - Go to: https://console.groq.com/
   - Get free API key
   - Cost: Free tier available! (Great for learning)

### Setting Up Your .env File

1. Copy the template:
```bash
cp .env.example .env
```

2. Edit `.env` with your API keys:
```bash
nano .env  # or use any text editor
```

3. Add your keys:
```
OPENAI_API_KEY=sk-your-actual-key-here
ANTHROPIC_API_KEY=sk-ant-your-actual-key-here
GROQ_API_KEY=gsk_your-actual-key-here
```

4. Verify .env is in .gitignore (it should be!):
```bash
grep .env .gitignore
```

````python
# Load environment variables
from dotenv import load_dotenv
import os

# Load .env file
load_dotenv()

print("🔐 Checking API Key Configuration")
print("=" * 50)

# Check which API keys are configured
api_keys = {
    'OPENAI_API_KEY': os.getenv('OPENAI_API_KEY'),
    'ANTHROPIC_API_KEY': os.getenv('ANTHROPIC_API_KEY'),
    'GROQ_API_KEY': os.getenv('GROQ_API_KEY'),
    'HUGGINGFACE_TOKEN': os.getenv('HUGGINGFACE_TOKEN'),
}

configured_keys = []
missing_keys = []

for key_name, key_value in api_keys.items():
    if key_value and not key_value.startswith('your-'):
        # Show only first 10 and last 4 characters for security
        masked_key = f"{key_value[:10]}...{key_value[-4:]}"
        print(f"✅ {key_name:<25} Configured: {masked_key}")
        configured_keys.append(key_name)
    else:
        print(f"❌ {key_name:<25} Not configured")
        missing_keys.append(key_name)

print("=" * 50)

if len(configured_keys) >= 1:
    print(f"\n✅ You have {len(configured_keys)} API key(s) configured!")
    print("   You're ready to start making API calls.")
else:
    print("\n⚠️  No API keys configured yet.")
    print("   Please edit your .env file with at least one API key.")

if missing_keys:
    print(f"\n💡 Optional keys not configured: {', '.join(missing_keys)}")
    print("   You can add these later as needed.")
````

## 🧪 Step 4: Test Your First API Call

Let's make your first LLM API call! We'll start with whichever provider you configured.

````python
import time

print("🧪 Testing API Connections")
print("=" * 70)

# Test function for each provider
def test_openai():
    """Test OpenAI API connection"""
    try:
        from openai import OpenAI
        client = OpenAI(api_key=os.getenv('OPENAI_API_KEY'))
        
        start_time = time.time()
        response = client.chat.completions.create(
            model="gpt-3.5-turbo",
            messages=[{"role": "user", "content": "Say 'Hello from OpenAI!' in exactly 5 words."}],
            max_tokens=20
        )
        elapsed = time.time() - start_time
        
        message = response.choices[0].message.content
        tokens = response.usage.total_tokens
        cost = (tokens / 1000) * 0.0015  # GPT-3.5-turbo pricing
        
        print(f"✅ OpenAI (GPT-3.5)")
        print(f"   Response: {message}")
        print(f"   Tokens: {tokens} | Cost: ${cost:.6f} | Time: {elapsed:.2f}s")
        return True
    except Exception as e:
        print(f"❌ OpenAI: {str(e)[:100]}")
        return False

def test_anthropic():
    """Test Anthropic API connection"""
    try:
        import anthropic
        client = anthropic.Anthropic(api_key=os.getenv('ANTHROPIC_API_KEY'))
        
        start_time = time.time()
        response = client.messages.create(
            model="claude-3-5-sonnet-20241022",
            max_tokens=20,
            messages=[{"role": "user", "content": "Say 'Hello from Claude!' in exactly 5 words."}]
        )
        elapsed = time.time() - start_time
        
        message = response.content[0].text
        tokens = response.usage.input_tokens + response.usage.output_tokens
        cost = (response.usage.input_tokens / 1000) * 0.003 + (response.usage.output_tokens / 1000) * 0.015
        
        print(f"✅ Anthropic (Claude 3.5 Sonnet)")
        print(f"   Response: {message}")
        print(f"   Tokens: {tokens} | Cost: ${cost:.6f} | Time: {elapsed:.2f}s")
        return True
    except Exception as e:
        print(f"❌ Anthropic: {str(e)[:100]}")
        return False

def test_groq():
    """Test Groq API connection"""
    try:
        from groq import Groq
        client = Groq(api_key=os.getenv('GROQ_API_KEY'))
        
        start_time = time.time()
        response = client.chat.completions.create(
            model="mixtral-8x7b-32768",
            messages=[{"role": "user", "content": "Say 'Hello from Groq!' in exactly 5 words."}],
            max_tokens=20
        )
        elapsed = time.time() - start_time
        
        message = response.choices[0].message.content
        tokens = response.usage.total_tokens
        
        print(f"✅ Groq (Mixtral-8x7b)")
        print(f"   Response: {message}")
        print(f"   Tokens: {tokens} | Cost: ~$0 (free tier) | Time: {elapsed:.2f}s")
        return True
    except Exception as e:
        print(f"❌ Groq: {str(e)[:100]}")
        return False

# Test each provider
results = {
    'OpenAI': test_openai() if os.getenv('OPENAI_API_KEY') and not os.getenv('OPENAI_API_KEY').startswith('your-') else None,
    'Anthropic': test_anthropic() if os.getenv('ANTHROPIC_API_KEY') and not os.getenv('ANTHROPIC_API_KEY').startswith('your-') else None,
    'Groq': test_groq() if os.getenv('GROQ_API_KEY') and not os.getenv('GROQ_API_KEY').startswith('your-') else None,
}

print("=" * 70)

# Summary
working = [k for k, v in results.items() if v is True]
failed = [k for k, v in results.items() if v is False]
not_configured = [k for k, v in results.items() if v is None]

if working:
    print(f"\n🎉 SUCCESS! {len(working)} provider(s) working: {', '.join(working)}")
    print("   You're ready to start building AI applications!")
else:
    print("\n⚠️  No providers are working yet.")
    print("   Please check your API keys in .env file.")

if failed:
    print(f"\n❌ Failed: {', '.join(failed)}")
    print("   Check error messages above for details.")

if not_configured:
    print(f"\n💡 Not configured: {', '.join(not_configured)}")
    print("   You can add these API keys later.")
````

````python
# Solution implementation
def check_environment_health() -> dict:
    """
    Check if the environment is properly set up for AI development.
    
    Returns:
        dict: Health check results with status for each component
    """
    import sys
    import os
    
    results = {
        'python_version': False,
        'packages_installed': False,
        'api_keys_configured': False,
        'ready': False
    }
    
    # Check Python version
    if sys.version_info >= (3, 11):
        results['python_version'] = True
    
    # Check packages
    required = ['openai', 'anthropic', 'langchain', 'pandas']
    installed = []
    for pkg in required:
        try:
            __import__(pkg)
            installed.append(pkg)
        except ImportError:
            pass
    
    if len(installed) == len(required):
        results['packages_installed'] = True
    
    # Check API keys
    api_keys = ['OPENAI_API_KEY', 'ANTHROPIC_API_KEY', 'GROQ_API_KEY']
    configured = sum(1 for key in api_keys if os.getenv(key) and not os.getenv(key).startswith('your-'))
    
    if configured >= 1:
        results['api_keys_configured'] = True
    
    # Overall readiness
    results['ready'] = all([results['python_version'], results['packages_installed'], results['api_keys_configured']])
    
    # Print report
    print("🏥 Environment Health Check")
    print("=" * 50)
    print(f"Python 3.11+:        {'✅' if results['python_version'] else '❌'}")
    print(f"Packages installed:  {'✅' if results['packages_installed'] else '❌'}")
    print(f"API keys configured: {'✅' if results['api_keys_configured'] else '❌'}")
    print("=" * 50)
    print(f"Overall status:      {'✅ READY' if results['ready'] else '❌ NOT READY'}")
    
    return results

# Run the health check
health = check_environment_health()
````

## ⚠️ Common Pitfalls & Troubleshooting

### Problem 1: "ModuleNotFoundError"
```python
ModuleNotFoundError: No module named 'openai'
```
**Solution:**
```bash
# Make sure you're in your virtual environment
source venv/bin/activate
pip install openai
```

### Problem 2: "AuthenticationError"
```
openai.AuthenticationError: Invalid API key
```
**Solutions:**
1. Check .env file has correct API key format
2. Restart Jupyter kernel after editing .env
3. Verify you copied the full API key (they're long!)

### Problem 3: Jupyter not finding .env file
**Solution:**
```python
# Specify the path explicitly
from dotenv import load_dotenv
load_dotenv('.env')  # or full path
```

### Problem 4: Wrong Python version
```bash
# Create venv with specific Python version
python3.11 -m venv venv
```

### Problem 5: "Rate limit exceeded"
**Solutions:**
1. Wait a minute and try again
2. Use Groq for testing (higher rate limits)
3. Implement exponential backoff (we'll learn this!)

### Problem 6: Slow API responses
**Normal latencies:**
- Groq: 0.2-0.5s ✅
- OpenAI: 1-3s ✅
- Anthropic: 1-2s ✅

If slower:
1. Check your internet connection
2. Try a different provider
3. Check provider status pages

## 🏭 Production Considerations

### In Real Production Environments:

1. **Secrets Management**
   - Use AWS Secrets Manager, Azure Key Vault, or HashiCorp Vault
   - Never store secrets in code or .env files in production
   - Rotate API keys regularly (every 90 days)

2. **Environment Isolation**
   - Separate environments: dev, staging, production
   - Different API keys for each environment
   - Different rate limits and budgets

3. **Monitoring**
   - Track API usage and costs in real-time
   - Set up alerts for unusual activity
   - Log all API calls for debugging

4. **Docker Containers**
   - Package your environment in Docker
   - Ensures consistency across team

5. **CI/CD Integration**
   - Automated testing of environment setup
   - Automatic dependency updates
   - We'll cover this in Module 18!

- Set hard limits on API spending
