## 🎯 Basic Synthetic Generation

````python
def generate_synthetic_example(
    task_description: str,
    example_format: Dict,
    context: str = None
) -> Dict:
    """Generate a single synthetic training example."""
    
    prompt = f"""Generate a training example for this task:

Task: {task_description}

Format:
{json.dumps(example_format, indent=2)}

{f'Context: {context}' if context else ''}

Generate ONE example in the exact JSON format above. Be creative and realistic.

JSON:"""
    
    response = client.chat.completions.create(
        model="gpt-4",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.8  # Higher for variety
    )
    
    try:
        example = json.loads(response.choices[0].message.content)
        return example
    except:
        return {"error": "Failed to parse"}

# Example usage
task = "Customer support for a tech company"
format_example = {
    "user_query": "How do I reset my password?",
    "assistant_response": "To reset your password: 1. Go to login page 2. Click 'Forgot Password'..."
}

print("🎯 Generating synthetic example...\n")

synthetic_example = generate_synthetic_example(task, format_example)

print("Generated example:")
print(json.dumps(synthetic_example, indent=2))

print("\n💡 Generate many variations by calling this function repeatedly")
````

## 🎲 Diverse Data Generation

````python
class SyntheticDataGenerator:
    """Generate diverse synthetic training data."""
    
    def __init__(self, seed_examples: List[Dict] = None):
        self.seed_examples = seed_examples or []
        self.client = OpenAI(api_key=os.getenv('OPENAI_API_KEY'))
    
    def generate_batch(
        self,
        task_description: str,
        num_examples: int,
        diversity_instructions: List[str] = None
    ) -> List[Dict]:
        """Generate a batch of diverse examples."""
        
        if diversity_instructions is None:
            diversity_instructions = [
                "Make this example simple and straightforward",
                "Include an edge case or unusual scenario",
                "Make this example more complex with multiple parts",
                "Include a common user mistake",
                "Make this example very specific and detailed",
            ]
        
        examples = []
        
        for i in range(num_examples):
            # Rotate through diversity instructions
            instruction = diversity_instructions[i % len(diversity_instructions)]
            
            prompt = f"""Generate a training example for: {task_description}

Diversity instruction: {instruction}

{"Example format (based on seed data):" if self.seed_examples else ""}
{json.dumps(self.seed_examples[0], indent=2) if self.seed_examples else ""}

Generate ONE unique example as JSON:
{{
  "user_input": "...",
  "assistant_output": "..."
}}

JSON:"""
            
            response = self.client.chat.completions.create(
                model="gpt-4",
                messages=[{"role": "user", "content": prompt}],
                temperature=0.9  # High diversity
            )
            
            try:
                example = json.loads(response.choices[0].message.content)
                examples.append(example)
            except:
                continue
        
        return examples
    
    def validate_quality(self, examples: List[Dict]) -> Dict:
        """Check quality of generated examples."""
        
        issues = []
        
        # Check for duplicates
        inputs = [ex.get('user_input', '') for ex in examples]
        unique_inputs = set(inputs)
        
        if len(unique_inputs) < len(inputs):
            issues.append(f"Found {len(inputs) - len(unique_inputs)} duplicate inputs")
        
        # Check lengths
        short_examples = sum(1 for ex in examples if len(str(ex)) < 50)
        if short_examples > len(examples) * 0.2:
            issues.append(f"{short_examples} examples are too short")
        
        return {
            'valid': len(issues) == 0,
            'issues': issues,
            'total': len(examples),
            'unique': len(unique_inputs),
        }

# Example
print("🎲 Generating diverse dataset...\n")

generator = SyntheticDataGenerator()

task = "Customer support for password resets"

# Generate 5 diverse examples
synthetic_data = generator.generate_batch(task, num_examples=5)

print(f"Generated {len(synthetic_data)} examples:\n")
for i, example in enumerate(synthetic_data[:2], 1):  # Show first 2
    print(f"Example {i}:")
    print(json.dumps(example, indent=2))
    print()

# Validate
validation = generator.validate_quality(synthetic_data)
print(f"Quality check: {validation}")
````

## ✅ Quality Control

````python
def llm_quality_check(example: Dict) -> Dict:
    """Use LLM to evaluate synthetic data quality."""
    
    prompt = f"""Evaluate this synthetic training example:

{json.dumps(example, indent=2)}

Rate the quality on these criteria (1-5):
- Realistic: Does it look like real user data?
- Helpful: Is the response actually helpful?
- Format: Is it properly formatted?
- Clarity: Is it clear and understandable?

Respond in JSON:
{{
  "realistic": <1-5>,
  "helpful": <1-5>,
  "format": <1-5>,
  "clarity": <1-5>,
  "overall": <average>,
  "issues": ["issue1", "issue2"],
  "keep": true/false
}}"""
    
    response = client.chat.completions.create(
        model="gpt-4",
        messages=[{"role": "user", "content": prompt}],
        temperature=0
    )
    
    try:
        return json.loads(response.choices[0].message.content)
    except:
        return {"error": "Failed to parse"}

def filter_high_quality(
    examples: List[Dict],
    min_score: float = 3.5
) -> List[Dict]:
    """Filter synthetic data by quality."""
    
    high_quality = []
    
    for example in examples:
        quality = llm_quality_check(example)
        
        if quality.get('overall', 0) >= min_score:
            high_quality.append(example)
    
    return high_quality

print("✅ Quality Control Example\n")
print("""
# Generate 100 examples
synthetic_data = generator.generate_batch(task, num_examples=100)

# Filter for quality
high_quality = filter_high_quality(synthetic_data, min_score=4.0)

print(f"Generated: {len(synthetic_data)}")
print(f"High quality: {len(high_quality)} ({len(high_quality)/len(synthetic_data)*100:.0f}%)")

Output:
Generated: 100
High quality: 73 (73%)

💡 Expect 70-80% of synthetic data to be high quality
""")
````

## 🔄 Mixing Real and Synthetic Data

````python
def create_mixed_dataset(
    real_examples: List[Dict],
    target_size: int,
    real_ratio: float = 0.2
) -> List[Dict]:
    """Create dataset mixing real and synthetic data."""
    
    num_real = min(len(real_examples), int(target_size * real_ratio))
    num_synthetic = target_size - num_real
    
    print(f"Creating mixed dataset:")
    print(f"  Real examples: {num_real} ({real_ratio*100:.0f}%)")
    print(f"  Synthetic: {num_synthetic} ({(1-real_ratio)*100:.0f}%)")
    print(f"  Total: {target_size}\n")
    
    # Use real examples as seeds
    generator = SyntheticDataGenerator(seed_examples=real_examples)
    
    # Generate synthetic data
    task = "Similar to the seed examples provided"
    synthetic_data = generator.generate_batch(task, num_synthetic)
    
    # Mix
    mixed_dataset = real_examples[:num_real] + synthetic_data
    
    # Shuffle
    random.shuffle(mixed_dataset)
    
    return mixed_dataset

# Example
print("🔄 Mixed Dataset Strategy\n")
print("""
# Start with small real dataset
real_examples = load_real_data()  # 50 examples

# Create mixed dataset
mixed_dataset = create_mixed_dataset(
    real_examples=real_examples,
    target_size=500,
    real_ratio=0.2  # 20% real, 80% synthetic
)

Output:
Creating mixed dataset:
  Real examples: 50 (20%)
  Synthetic: 450 (80%)
  Total: 500

💡 Recommended ratios:
  - 20% real, 80% synthetic: Good balance
  - 50% real, 50% synthetic: High quality
  - 10% real, 90% synthetic: Bootstrap mode
""")
````

## ✅ Summary

### Synthetic Data Workflow:

```python
# 1. Start with seed examples (even just 5-10)
seed_examples = [
    {"user": "...", "assistant": "..."},
    ...
]

# 2. Generate synthetic data
generator = SyntheticDataGenerator(seed_examples=seed_examples)
synthetic = generator.generate_batch(task, num_examples=500)

# 3. Quality filter
high_quality = filter_high_quality(synthetic, min_score=4.0)

# 4. Mix with real data
final_dataset = seed_examples + high_quality

# 5. Train model
fine_tune(final_dataset)
```
