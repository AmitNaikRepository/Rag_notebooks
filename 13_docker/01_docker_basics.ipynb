## 🐳 Docker Fundamentals

### Key Concepts:

```
Image:
  - Blueprint for container
  - Contains code + dependencies
  - Immutable
  - Example: python:3.11-slim

Container:
  - Running instance of image
  - Isolated environment
  - Has own filesystem, network
  - Example: Your LLM API running

Dockerfile:
  - Recipe to build image
  - Text file with instructions
  - Version controlled
```

## 🚀 Running LLM App in Docker

### Basic Commands:

```bash
# Build image
docker build -t llm-api:latest .

# Run container
docker run -p 8000:8000 llm-api:latest

# Run with environment variables
docker run -e OPENAI_API_KEY=sk-... llm-api

# Run in background
docker run -d llm-api

# View running containers
docker ps

# View logs
docker logs <container-id>

# Stop container
docker stop <container-id>

# Remove container
docker rm <container-id>
```

## 💾 Volumes

```bash
# Mount local directory
docker run -v $(pwd)/data:/app/data llm-api

# Named volume
docker volume create llm-cache
docker run -v llm-cache:/app/cache llm-api

# Read-only mount
docker run -v $(pwd)/config:/app/config:ro llm-api
```

### Use cases:
- **Data persistence**: Database files
- **Configuration**: .env files
- **Cache**: Model weights, embeddings
- **Logs**: Application logs

## ✅ Summary

**Docker benefits for LLM apps:**
- ✅ Consistent environments
- ✅ Easy deployment
- ✅ Isolation from host
- ✅ Reproducible builds

**Basic workflow:**
```bash
1. Write Dockerfile
2. Build image: docker build -t app .
3. Run container: docker run -p 8000:8000 app
4. Deploy anywhere
```
