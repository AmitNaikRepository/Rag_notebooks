## 🚀 Basic Example

```python
from ray import serve
import ray

# Initialize
ray.init()
serve.start()

# Define deployment
@serve.deployment(num_replicas=2)
class LLMModel:
    def __init__(self):
        from transformers import pipeline
        self.model = pipeline(
            "text-generation",
            model="gpt2"
        )
    
    async def __call__(self, request):
        prompt = await request.json()
        result = self.model(
            prompt["text"],
            max_length=100
        )
        return {"response": result[0]["generated_text"]}

# Deploy
LLMModel.deploy()

# Use
import requests
response = requests.post(
    "http://localhost:8000/",
    json={"text": "Hello"}
)
```
