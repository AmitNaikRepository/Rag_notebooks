```python
from sentence_transformers import SentenceTransformer, util
import numpy as np
from scipy.spatial.distance import cosine, euclidean
from typing import List, Tuple
import pandas as pd

model = SentenceTransformer('all-MiniLM-L6-v2')
print("✅ Setup complete")
```

## 📏 Understanding Distance Metrics

### Cosine Similarity (Most Popular)
- Measures angle between vectors
- Range: -1 (opposite) to 1 (identical)
- Ignores magnitude, focuses on direction
- **Use**: Text similarity, search

### Euclidean Distance
- Straight-line distance between vectors
- Range: 0 (identical) to infinity
- Considers magnitude
- **Use**: Spatial data, clustering

### Dot Product
- Simple multiplication
- Range: -infinity to infinity
- Fast to compute
- **Use**: Normalized vectors

```python
# Example texts
texts = [
    "I love programming in Python",
    "Python is my favorite language",  # Very similar
    "I enjoy cooking Italian food",    # Different
]

# Create embeddings
embeddings = model.encode(texts)

# Cosine similarity
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(embeddings)

print("Cosine Similarity Matrix:")
print("(1.0 = identical, 0.0 = unrelated)\n")

df = pd.DataFrame(
    similarity_matrix,
    index=[f"Text {i+1}" for i in range(len(texts))],
    columns=[f"Text {i+1}" for i in range(len(texts))]
)

print(df.round(3))

print(f"\nText 1 vs Text 2: {similarity_matrix[0][1]:.3f} (very similar!)")
print(f"Text 1 vs Text 3: {similarity_matrix[0][2]:.3f} (different)")
```

## 🔍 Semantic Search Implementation

```python
class SemanticSearch:
    """Simple semantic search engine."""
    
    def __init__(self, documents: List[str]):
        self.documents = documents
        self.model = SentenceTransformer('all-MiniLM-L6-v2')
        self.embeddings = self.model.encode(documents)
    
    def search(self, query: str, top_k: int = 3) -> List[Tuple[str, float]]:
        """Find most similar documents."""
        query_embedding = self.model.encode([query])[0]
        
        # Calculate similarities
        similarities = cosine_similarity(
            [query_embedding],
            self.embeddings
        )[0]
        
        # Get top k
        top_indices = np.argsort(similarities)[::-1][:top_k]
        
        results = [
            (self.documents[i], similarities[i])
            for i in top_indices
        ]
        
        return results

# Example: Search product descriptions
products = [
    "Wireless bluetooth headphones with noise cancellation",
    "USB-C charging cable 6 feet long",
    "Smartphone with 128GB storage and 5G",
    "Laptop computer with 16GB RAM",
    "Portable bluetooth speaker waterproof",
]

search_engine = SemanticSearch(products)

# Search
query = "headset for music"
results = search_engine.search(query, top_k=3)

print(f"Query: '{query}'\n")
print("Top Results:")
for i, (doc, score) in enumerate(results, 1):
    print(f"{i}. [{score:.3f}] {doc}")
```

## 📊 Visualizing Similarity

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create similarity heatmap
texts_for_viz = [
    "dog",
    "puppy",
    "cat",
    "kitten",
    "car",
    "vehicle"
]

embeddings_viz = model.encode(texts_for_viz)
sim_matrix = cosine_similarity(embeddings_viz)

plt.figure(figsize=(8, 6))
sns.heatmap(
    sim_matrix,
    annot=True,
    fmt='.2f',
    xticklabels=texts_for_viz,
    yticklabels=texts_for_viz,
    cmap='YlOrRd'
)
plt.title('Semantic Similarity Heatmap')
plt.tight_layout()
plt.show()

print("Notice:")
print("  • dog ↔ puppy: High similarity")
print("  • cat ↔ kitten: High similarity")
print("  • dog ↔ car: Low similarity")
```

## ✅ Summary

### Key Takeaways:
- 📐 **Cosine similarity**: Best for text (0-1 range)
- 🎯 **Semantic search**: Find meaning, not keywords
- ⚡ **Efficient**: Matrix operations are fast
- 📊 **Visualize**: Heatmaps show relationships

### When to Use:
- **Cosine**: Text similarity, search (MOST COMMON)
- **Euclidean**: Clustering, spatial data
- **Dot product**: Fast approximation
