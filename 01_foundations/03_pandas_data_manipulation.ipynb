## 📊 Why Pandas for AI?

**Common AI tasks:**
- Load training data
- Clean and preprocess text
- Analyze model outputs
- Track experiments
- Prepare datasets

```python
# Example: LLM training data
data = {
    'prompt': [
        'Summarize this article',
        'Translate to Spanish',
        'Classify sentiment',
    ],
    'response': [
        'Article summary here...',
        'Traducción aquí...',
        'Positive',
    ],
    'tokens': [150, 80, 20],
    'cost': [0.0045, 0.0024, 0.0006]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Common operations
print("\nBasic stats:")
print(df[['tokens', 'cost']].describe())

print("\nTotal cost:")
print(f"${df['cost'].sum():.4f}")

# Filter
expensive = df[df['cost'] > 0.003]
print(f"\nExpensive calls: {len(expensive)}")
```

## 🔄 Data Cleaning for LLMs

```python
# Sample messy text data
messy_data = pd.DataFrame({
    'text': [
        '  Text with spaces  ',
        'UPPERCASE TEXT',
        'Text\nwith\nnewlines',
        None,  # Missing value
    ]
})

# Clean it
clean = messy_data.copy()
clean['text'] = clean['text'].fillna('')  # Fill missing
clean['text'] = clean['text'].str.strip()  # Remove spaces
clean['text'] = clean['text'].str.replace('\n', ' ')  # Remove newlines
clean['text'] = clean['text'].str.lower()  # Lowercase

print("Before:")
print(messy_data)
print("\nAfter:")
print(clean)
```
