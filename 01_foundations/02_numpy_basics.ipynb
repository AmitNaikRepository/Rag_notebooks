## 📊 Why NumPy for AI?

**Embeddings are NumPy arrays:**
```python
embedding = np.array([0.1, 0.2, 0.3, ...])  # 384 dimensions
```

**Why not Python lists?**
- ⚡ 50-100x faster
- 💾 Uses less memory
- 🎯 Optimized for math operations
- 📊 Matrix operations built-in

````python
# Speed comparison: Lists vs NumPy
size = 1000000

# Python list
python_list = list(range(size))
start = time.time()
result_list = [x * 2 for x in python_list]
time_list = time.time() - start

# NumPy array
numpy_array = np.arange(size)
start = time.time()
result_numpy = numpy_array * 2
time_numpy = time.time() - start

print(f"Python list: {time_list*1000:.2f}ms")
print(f"NumPy array: {time_numpy*1000:.2f}ms")
print(f"\nNumPy is {time_list/time_numpy:.1f}x faster! ⚡")
````

## 🎯 NumPy for Embeddings

````python
# Create sample embeddings
embedding1 = np.random.rand(384)  # 384-dimensional vector
embedding2 = np.random.rand(384)

# Cosine similarity (critical for semantic search)
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

similarity = cosine_similarity(embedding1, embedding2)
print(f"Cosine similarity: {similarity:.4f}")

# Euclidean distance
distance = np.linalg.norm(embedding1 - embedding2)
print(f"Euclidean distance: {distance:.4f}")
````

## 📈 Array Operations You'll Use Daily

````python
# Create arrays
arr = np.array([1, 2, 3, 4, 5])
print(f"Array: {arr}")
print(f"Shape: {arr.shape}")
print(f"Type: {arr.dtype}")

# Common operations
print(f"\nMean: {arr.mean()}")
print(f"Std: {arr.std()}")
print(f"Min: {arr.min()}, Max: {arr.max()}")
print(f"Sum: {arr.sum()}")
````

````python
# Matrix operations (for batch embeddings)
embeddings = np.random.rand(10, 384)  # 10 embeddings
print(f"Batch embeddings shape: {embeddings.shape}")
print(f"Memory: {embeddings.nbytes / 1024:.2f} KB")

# Normalize embeddings (common preprocessing)
norms = np.linalg.norm(embeddings, axis=1, keepdims=True)
normalized = embeddings / norms
print(f"\nNormalized shape: {normalized.shape}")
print(f"First vector norm: {np.linalg.norm(normalized[0]):.4f}")  # Should be 1.0
````
