## ⚡ Async Programming (Critical for LLMs!)

```python
import asyncio
import time

# Synchronous - slow!
def slow_api_call(n: int) -> str:
    time.sleep(1)  # Simulate API delay
    return f"Result {n}"

# Asynchronous - fast!
async def fast_api_call(n: int) -> str:
    await asyncio.sleep(1)  # Non-blocking
    return f"Result {n}"

# Run multiple calls in parallel
async def main():
    start = time.time()
    
    # Parallel execution
    results = await asyncio.gather(
        fast_api_call(1),
        fast_api_call(2),
        fast_api_call(3)
    )
    
    elapsed = time.time() - start
    print(f"Completed 3 calls in {elapsed:.1f}s (parallel!)")
    return results

# Run async code
results = await main()
print("Results:", results)
```

## 🏷️ Type Hints & Pydantic (For Production Code)

```python
from typing import List, Dict, Optional
from pydantic import BaseModel, Field, validator

# Type hints - catch errors early
def calculate_cost(tokens: int, price_per_1k: float) -> float:
    return (tokens / 1000) * price_per_1k

# Pydantic - automatic validation
class LLMRequest(BaseModel):
    prompt: str = Field(..., min_length=1, max_length=2000)
    max_tokens: int = Field(default=100, ge=1, le=4000)
    temperature: float = Field(default=0.7, ge=0.0, le=2.0)
    
    @validator('prompt')
    def prompt_not_empty(cls, v):
        if not v.strip():
            raise ValueError('Prompt cannot be empty')
        return v

# Try it
try:
    request = LLMRequest(prompt="Test", max_tokens=50)
    print("✅ Valid request:", request.dict())
    
    # This will fail validation
    bad_request = LLMRequest(prompt="", max_tokens=5000)
except Exception as e:
    print(f"❌ Validation error: {e}")
```
