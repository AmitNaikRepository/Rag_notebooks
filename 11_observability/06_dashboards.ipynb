## 📊 Essential LLM Dashboard Panels

### 1. Request Rate
```promql
# Total requests per second
rate(llm_requests_total[5m])

# By model
sum by (model) (rate(llm_requests_total[5m]))
```

### 2. Error Rate
```promql
# Error percentage
sum(rate(llm_requests_total{status="error"}[5m]))
/
sum(rate(llm_requests_total[5m]))
* 100
```

### 3. Latency (Heatmap)
```promql
# P50, P95, P99 latency
histogram_quantile(0.50, rate(llm_latency_seconds_bucket[5m]))
histogram_quantile(0.95, rate(llm_latency_seconds_bucket[5m]))
histogram_quantile(0.99, rate(llm_latency_seconds_bucket[5m]))
```

### 4. Cost Tracking
```promql
# Cost per hour
rate(llm_cost_dollars[1h]) * 3600

# Cost by model
sum by (model) (rate(llm_cost_dollars[1h])) * 3600
```

### 5. Token Usage
```promql
# Tokens per second
rate(llm_tokens_total[5m])

# Prompt vs completion
sum by (type) (rate(llm_tokens_total[5m]))
```

### 6. Cache Performance
```promql
# Hit rate
rate(cache_hits_total[5m])
/
(rate(cache_hits_total[5m]) + rate(cache_misses_total[5m]))
* 100
```
