## 💾 Backup Strategies

### 1. pg_dump (Logical Backup)
```bash
# Backup single database
pg_dump llm_app > backup.sql

# Backup with compression
pg_dump llm_app | gzip > backup.sql.gz

# Restore
psql llm_app < backup.sql
```

### 2. Automated Backups
```bash
# Cron job (daily at 2 AM)
0 2 * * * pg_dump llm_app | \
  gzip > /backups/llm_app_$(date +\%Y\%m\%d).sql.gz
```

### 3. AWS RDS Automated Backups
- Automatic daily snapshots
- Point-in-time recovery
- Multi-AZ replication
