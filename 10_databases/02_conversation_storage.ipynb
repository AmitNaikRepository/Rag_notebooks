## 💬 Conversation Manager

```python
class ConversationManager:
    def __init__(self, db):
        self.db = db
    
    def create_conversation(self, user_id, title):
        query = '''
            INSERT INTO conversations (user_id, title)
            VALUES (%s, %s)
            RETURNING id
        '''
        return self.db.execute(query, (user_id, title))
    
    def add_message(self, conv_id, role, content, tokens):
        query = '''
            INSERT INTO messages 
            (conversation_id, role, content, tokens)
            VALUES (%s, %s, %s, %s)
        '''
        self.db.execute(query, (conv_id, role, content, tokens))
    
    def get_messages(self, conv_id, limit=50):
        query = '''
            SELECT role, content, created_at
            FROM messages
            WHERE conversation_id = %s
            ORDER BY created_at ASC
            LIMIT %s
        '''
        return self.db.query(query, (conv_id, limit))
```

## ✅ Summary

**Best practices:**
- Separate conversations and messages
- Use pagination for long conversations
- Index conversation_id for fast retrieval
