## 🦜 LangChain Agents

````python
print("🦜 LangChain Agent Example\n")

print("""
from langchain.agents import AgentExecutor, create_openai_functions_agent
from langchain_openai import ChatOpenAI
from langchain.tools import Tool
from langchain import hub

# Define tools
def search(query: str) -> str:
    return f"Search results for: {query}"

def calculate(expression: str) -> str:
    return str(eval(expression))

tools = [
    Tool(
        name="Search",
        func=search,
        description="Search for information"
    ),
    Tool(
        name="Calculate",
        func=calculate,
        description="Perform calculations"
    )
]

# Create agent
llm = ChatOpenAI(model="gpt-4", temperature=0)
prompt = hub.pull("hwchase17/openai-functions-agent")

agent = create_openai_functions_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)

# Run agent
result = agent_executor.invoke({
    "input": "What is 25 * 4?"
})

print(result["output"])

✅ Key Features:
  - Built-in agent types (ReAct, OpenAI Functions, etc.)
  - Large tool library
  - Memory management
  - Chain composition
  - Production-ready
""")
````

## 🦙 LlamaIndex Agents

````python
print("🦙 LlamaIndex Agent Example\n")

print("""
from llama_index.core.agent import ReActAgent
from llama_index.llms.openai import OpenAI
from llama_index.core.tools import FunctionTool

# Define tools
def multiply(a: int, b: int) -> int:
    \"\"\"Multiply two numbers.\"\"\" 
    return a * b

def add(a: int, b: int) -> int:
    \"\"\"Add two numbers.\"\"\"  
    return a + b

# Create tools
multiply_tool = FunctionTool.from_defaults(fn=multiply)
add_tool = FunctionTool.from_defaults(fn=add)

# Create agent
llm = OpenAI(model="gpt-4")
agent = ReActAgent.from_tools(
    [multiply_tool, add_tool],
    llm=llm,
    verbose=True
)

# Run agent
response = agent.chat("What is (5 + 3) * 2?")
print(response)

✅ Key Features:
  - Great for data agents
  - RAG integration
  - Query engines
  - Data connectors
  - Simple API
""")
````

## 👥 CrewAI

````python
print("👥 CrewAI Example\n")

print("""
from crewai import Agent, Task, Crew

# Define agents
researcher = Agent(
    role='Researcher',
    goal='Research and gather information',
    backstory='Expert researcher with attention to detail',
    verbose=True
)

writer = Agent(
    role='Writer',
    goal='Write compelling content',
    backstory='Talented writer who creates engaging narratives',
    verbose=True
)

# Define tasks
research_task = Task(
    description='Research AI agents',
    agent=researcher,
    expected_output='Comprehensive research notes'
)

write_task = Task(
    description='Write article about AI agents',
    agent=writer,
    expected_output='Well-written article'
)

# Create crew
crew = Crew(
    agents=[researcher, writer],
    tasks=[research_task, write_task],
    verbose=True
)

# Run crew
result = crew.kickoff()
print(result)

✅ Key Features:
  - Role-based agents
  - Easy multi-agent setup
  - Task delegation
  - Sequential/parallel execution
  - Intuitive API
""")
````

## 🔄 AutoGen

````python
print("🔄 AutoGen Example\n")

print("""
from autogen import AssistantAgent, UserProxyAgent

# Configure LLM
config_list = [{
    "model": "gpt-4",
    "api_key": os.getenv("OPENAI_API_KEY")
}]

# Create assistant
assistant = AssistantAgent(
    name="assistant",
    llm_config={"config_list": config_list}
)

# Create user proxy (can execute code)
user_proxy = UserProxyAgent(
    name="user_proxy",
    human_input_mode="NEVER",
    code_execution_config={"work_dir": "coding"}
)

# Start conversation
user_proxy.initiate_chat(
    assistant,
    message="Plot a chart of NVIDIA and TESLA stock price changes."
)

✅ Key Features:
  - Multi-agent conversations
  - Code execution
  - Human-in-the-loop
  - Group chat
  - Teaching/learning agents
""")
````

## ✅ Summary

### Framework Selection Guide:

**Choose LangChain when:**
- Need comprehensive features
- Building complex chains
- Large community support needed
- Production-ready system

**Choose LlamaIndex when:**
- Focus on data/RAG
- Need indexing capabilities
- Simple agent needs
- Clean API preferred

**Choose CrewAI when:**
- Multi-agent teams
- Role-based agents
- Simple workflows
- Quick prototyping

**Choose AutoGen when:**
- Multi-agent conversations
- Code execution needed
- Teaching/learning agents
- Group chat scenarios

**Choose Custom when:**
- Full control needed
- Minimal dependencies
- Production critical
- Unique requirements

### Pros & Cons:

**Frameworks:**
```python
✅ Pros:
- Fast prototyping
- Pre-built tools
- Community support
- Less code

❌ Cons:
- Learning curve
- Dependency hell
- Abstraction overhead
- Version conflicts
- Harder debugging
```

**Custom:**
```python
✅ Pros:
- Full control
- Easy debugging
- No dependencies
- Production-ready
- Understand everything

❌ Cons:
- More code
- Build from scratch
- More time
```

### Migration Path:

```
1. Learn fundamentals (Custom)
   ↓
2. Prototype with framework (CrewAI/LlamaIndex)
   ↓
3. Scale with framework (LangChain)
   ↓
4. Production with custom (if needed)
```

### Best Practices:

**1. Start Simple**
- Master custom implementation first
- Understand core concepts
- Then add frameworks

**2. Don't Over-Framework**
- Use frameworks for what they're good at
- Don't force-fit everything
- Custom is often simpler

**3. Pin Versions**
```python
# requirements.txt
langchain==0.1.0  # Pin exact versions!

```

**4. Keep It Simple**
- Prefer simple over clever
- Readable > concise
- Explicit > implicit
