## 📖 System Prompts Explained

**Message Types:**
```python
system:  Sets the AI's behavior/role (persistent)
user:    The user's input (changes each time)
assistant: The AI's previous responses (context)
```

**Example:**
```
System: "You are a helpful Python tutor"
User: "What is a list?"
Assistant: "A list is..."
```

**Why use system prompts?**
- ✅ Consistent tone across conversations
- ✅ Define rules and constraints
- ✅ Set expertise level
- ✅ Control output format

````python
import os
from dotenv import load_dotenv
from groq import Groq

load_dotenv()
client = Groq(api_key=os.getenv('GROQ_API_KEY'))

def call_with_system(system_prompt, user_message):
    response = client.chat.completions.create(
        model="mixtral-8x7b-32768",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_message}
        ],
        temperature=0.7,
        max_tokens=150
    )
    return response.choices[0].message.content

print("✅ Setup complete")
````

## 🎯 System Prompt Examples

````python
user_query = "Explain machine learning"

# Example 1: Expert mode
system1 = "You are a senior ML engineer. Use technical terms."
response1 = call_with_system(system1, user_query)
print("Expert Mode:")
print(response1)
print("\n" + "="*60 + "\n")

# Example 2: Beginner mode
system2 = "You are teaching a 10-year-old. Use simple words and analogies."
response2 = call_with_system(system2, user_query)
print("Beginner Mode:")
print(response2)
print("\n" + "="*60 + "\n")

# Example 3: Concise mode
system3 = "Answer in exactly 2 sentences. Be extremely concise."
response3 = call_with_system(system3, user_query)
print("Concise Mode:")
print(response3)
````

## 📚 Production System Prompt Templates

````python
SYSTEM_PROMPTS = {
    'customer_support': """You are a helpful customer support agent.
    
Rules:
- Be polite and professional
- If you don't know, say so and offer to escalate
- Keep responses under 100 words
- Never make promises about refunds or policies
""",
    
    'code_reviewer': """You are an expert code reviewer.
    
Focus on:
- Security vulnerabilities
- Performance issues
- Code readability
- Best practices

Provide:
- Specific suggestions
- Severity rating (Critical/High/Medium/Low)
- Example fixes
""",
    
    'json_api': """You are a JSON API. Always respond with valid JSON only.
    
Rules:
- No explanatory text before or after JSON
- Use double quotes for strings
- Include 'status' and 'data' fields
- If error, set status='error' and include 'message'
    
Example: {"status": "success", "data": {...}}
""",
}

# Test JSON API system prompt
response = call_with_system(
    SYSTEM_PROMPTS['json_api'],
    "Extract name and email from: 'Contact John at john@email.com'"
)

print("JSON API Response:")
print(response)
````
